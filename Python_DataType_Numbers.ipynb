```python
#Numbers (int, float, complex)

#Introduction

Numbers are one of the fundamental data types in Python. Python supports integers, floating-point numbers, and complex numbers.

#Key Points and Definition

int: Whole numbers, e.g., 10, -3

float: Decimal numbers, e.g., 3.14, -0.001

complex: Numbers with a real and imaginary part, e.g., 2 + 3j

#Common Functions/Methods

abs(x) – Absolute value

round(x) – Rounds to nearest integer

pow(x, y) – Returns x raised to the power y
Example
x = 5       # int
y = 3.2     # float
z = 2 + 3j  # complex
print(abs(-10))       # 10
print(round(2.6))     # 3
```
