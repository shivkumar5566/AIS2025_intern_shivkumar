```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df['age'] = df['age'].fillna(0)  # fill age with 0
df['embarked'] = df['embarked'].fillna('Unknown')  # fill embarked with 'Unknown'
```

```python
df_cleaned = df.dropna()
```

```python
df = df.dropna(subset=['age'])  # drops only rows where age is missing
```
