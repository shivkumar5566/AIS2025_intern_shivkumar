```python
import seaborn as sns

# Set up the figure
fig, axs = plt.subplots(3, 2, figsize=(16, 18))
axs = axs.flatten()

# 1. Violin Plot
sns.violinplot(data=tips, x="day", y="total_bill", ax=axs[0])
axs[0].set_title("1. Violin Plot of Total Bill by Day")

# 2. Bar Plot
sns.barplot(data=tips, x="day", y="total_bill", ax=axs[1])
axs[1].set_title("2. Bar Plot of Average Total Bill by Day")

# 3. Count Plot
sns.countplot(data=tips, x="day", ax=axs[2])
axs[2].set_title("3. Count Plot of Days")

# 4. Strip Plot
sns.stripplot(data=tips, x="day", y="total_bill", jitter=True, ax=axs[3])
axs[3].set_title("4. Strip Plot of Total Bill by Day")

# 5. Swarm Plot
sns.swarmplot(data=tips, x="day", y="total_bill", ax=axs[4])
axs[4].set_title("5. Swarm Plot of Total Bill by Day")

# Hide the last empty subplot
fig.delaxes(axs[5])

# Show the plots
plt.tight_layout()
plt.show()
```
