```python
# Matplotlib  library
import matplotlib.pyplot as plt
```

```python
# 1. Plot Chart

# Data
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [8, 5, 9, 4, 9, 3, 1, 7]

# Create plot
plt.plot(x, y)
plt.title("Plot Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
#   Plot Chart

# Data
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [8, 5, 9, 4, 9, 3, 1, 7]

# Create plot
plt.plot(x, y)
plt.title("Plot Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(False)
plt.show()
```

```python
# 2. Bar Chart

# Data
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [8, 5, 9, 4, 9, 3, 1, 7]

# Create Bar chart
plt.bar(x, y)
plt.title("Bar Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python
# 3. Scatter Chart

# Data
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [8, 5, 9, 4, 9, 3, 1, 7]

# Create Scatter chart
plt.scatter(x, y, color='red')
plt.title("Scater plot Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
# 4.Pie Chart

#Data
Apps=[ 'whatsapp', 'instagram', 'facebook', 'snapchat']
sizes=[75, 64, 88, 58]

# Plot
plt.pie(sizes, autopct='%1.1f%%', startangle=140)
plt.title("Apps size")
plt.axis('equal') 
plt.show()
```

```python
# 5 Multiple Lines in One Plot

# Data
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 6, 4, 3, 2]

#Plot

plt.plot(x, y1, label='Squared')
plt.plot(x, y2, label='Linear')
plt.title("Multiple Lines")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.show()
```
