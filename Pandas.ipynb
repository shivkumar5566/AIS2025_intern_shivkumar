```python
# 2. Importing pandas:

import pandas as pd
```

```python
#3. Creating a DataFrame:

data = {
    'Name': ['shiva', 'pratik', 'ram', 'prashant', 'rohit', 'manohar', ],
    'Age': [25, 30, 35, 36, 18, 21,],
    'City': ['solapur', 'pune', 'Nagpur', 'solapur', 'pune', 'mumbai']
}

df = pd.DataFrame(data)
df
```

```python
# 4. Basic DataFrame Operations:
df.head()       # Show first 3 rows
df.info()       # Summary of the DataFrame
df.describe()   # Statistical summary of numeric columns
df['Age']       # Access a single column
df.iloc[0]      # Access first row
df[df['Age'] > 30]  # Filter rows
```
