```python
# String (str)

#Introduction

A string is a sequence of characters enclosed in quotes.

#Key Points and Definition

1)Immutable

2)Enclosed in single ('') or double ("") 

3)Supports indexing and slicing

#Indexing
Zero-based indexing, supports negative indexing

s = "Hello"
print(s[0])   # 'H'
print(s[-1])  # 'o'

#Common Methods

1)lower(), upper()

2)strip() – Removes whitespace

3)replace(old, new)

4)split() – Converts string to list

5)find(sub) – Returns index of substring

Example
```

```python
s = ("Hello World")
print(s.strip())          
print(s.replace("Hello", "Hi"))  
```
